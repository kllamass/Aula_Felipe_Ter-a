- Situação: Uma rede de supermercados registrou o faturamento diário de uma de suas unidades durante cinco dias úteis,
ver Tabela.

A direção estabeleceu que uma unidade apresenta:

- Bom desempenho: média diária igual ou superior a R$ 22.000;
- Desempenho regular: média entre R$ 19.000 e R$ 21.999,99;
• Desempenho crítico: média inferior a R$ 19.000.

Desenvolva um programa que permita analisar os dados da unidade e apresente:

- quantidade de dias analisados;
- faturamento total;
- faturamento médio diário;
- maior faturamento registrado;
- menor faturamento registrado;
- classificação do desempenho da unidade conforme os critérios da direção.


```python
faturamento = [18500 , 21300 , 17300 , 24600 , 27400]

total = sum(faturamento)
media = total/len(faturamento)

if media >= 22000:
    print("Bom desempenho")
elif media >= 1900 and media <= 21999.99:
    print("Desempenho regular")
else:
    print("desempenho critico")

qnt_dias = len(faturamento)
print(f"Foram analisados {qnt_dias} dias. O faturamento diário é de {media}. O maior faturamento foi {faturamento[4]}, enquanto o menor faturamento foi na quarta, um total de {faturamento[2]}")
```

independentemente da classificação obtida pela média.

- armazene as produções em uma lista;

# Pratica de calculo_variancia()

```python
def calcula_media(valores):
    media = sum(valores)/len(valores)
    return media
```

```python
lista = [12,18,15,21,14,16]

media = calcula_media(lista)
```

```python
lista = [12,18,15,21,14,16]

media = calcula_media(lista)



def calcula_variancia(lista):
    media = calcula_media(lista)
    soma_diferenca = 0

    for i in lista:
        diferenca = i - media
        diferenca_ao_quadrado = pow(diferenca,2) # calcula potencia tbm
        soma_diferenca += diferenca_ao_quadrado

    variancia = soma_diferenca/len(lista)
    return variancia
```

```python
calcula_variancia(lista)
```

```python
import math

def desvio_padrao(valores):
    variancia = calcula_variancia(valores)
    desvio = math.sqrt(variancia)
    return desvio

tempos = [12, 18, 15, 21, 14, 16]

resultado = desvio_padrao(tempos)
print(resultado)
```

# Calcular Quartis (socorro!)

```python
def calcular_quartis(dados):
    if not dados or len(dados) < 3:
        raise ValueError("A lista deve conter pelo menos 3 elementos.")

    dados_ordenados = sorted(dados)
    n = len(dados_ordenados)

    # Função auxiliar para calcular a mediana
    def mediana(lista):
        tam = len(lista)
        meio = tam // 2
        if tam % 2 == 0:
            return (lista[meio - 1] + lista[meio]) / 2.0
        else:
            return lista[meio]

    meio_geral = n // 2

    # Q2 é a mediana da lista inteira
    q2 = mediana(dados_ordenados)

    # Q1 e Q3 são as medianas da metade inferior e superior
    if n % 2 == 0:
        q1 = mediana(dados_ordenados[:meio_geral])
        q3 = mediana(dados_ordenados[meio_geral:])
    else:
        q1 = mediana(dados_ordenados[:meio_geral])
        q3 = mediana(dados_ordenados[meio_geral + 1:])

    return q1, q2, q3

# Exemplo de uso:
numeros = [7, 15, 36, 39, 40, 41]
q1, q2, q3 = calcular_quartis(numeros)
print(f"Q1: {q1} | Q2: {q2} | Q3: {q3}")
```

# IQR

```python
def calcular_iqr(valores):
    q1, q2, q3 = calcular_quartis(valores)
    iqr = q3 - q1
    return iqr

vendas = [42,18,35,27,50,31,22]
resultado = calcular_iqr(vendas)
print(resultado)
```

```python
def calcular_mediana(lista):
    if not lista:
        return None

    lista_ordenada = sorted(lista)
    n = len(lista_ordenada)
    meio = n // 2

    if n % 2 != 0:
        return lista_ordenada[meio]
    return (lista_ordenada[meio - 1] + lista_ordenada[meio]) / 2
```

# Parte 1: Processamento dos chamados
Desenvolva um programa em Python que:
1. Percorra a lista de chamados utilizando enumerate(), numerando os registros a partir de 1;
2. Converta as datas de abertura e fechamento, atualmente armazenadas como strings, para
objetos datetime;
3. Calcule o tempo de atendimento, em dias, de cada chamado;
4. Apresente, para cada chamado:

• número do registro;

• código;

• responsável;

• prioridade;

• tempo de atendimento em dias;

• situação do prazo.

5. Classifique cada chamado como:

• Dentro do prazo: tempo de atendimento menor ou igual a 3 dias;

• Fora do prazo: tempo de atendimento maior que 3 dias.

Uma possível apresentação para cada registro é:

1 - CH001 | Maria | Alta | 2 dias | Dentro do prazo

```python
for numero, chamado in enumerate(chamados, start=1):

    data_abertura = datetime.strptime(chamado["abertura"], "%d/%m/%Y")
    data_fechamento = datetime.strptime(chamado["fechamento"], "%d/%m/%Y")

    tempo_atendimento = (data_fechamento - data_abertura).days

    if tempo_atendimento <= 3:
        situacao = "Dentro do prazo"
    else:
        situacao = "Fora do prazo"

    print(f"{numero} - {chamado['codigo']} | {chamado['responsavel']} | {chamado['prioridade']} | {tempo_atendimento} dias | {situacao}")
```

#Parte 2: Resumo dos atendimentos
Após processar todos os chamados, apresente:
1. Quantidade total de chamados;

```python
dentro_prazo = []
fora_prazo = []
maior_tempo = 0
codigo_maior_tempo = ""
alta = []
media = []
baixa = []


for numero, chamado in enumerate(chamados, start=1):

    data_abertura = datetime.strptime(chamado["abertura"], "%d/%m/%Y")
    data_fechamento = datetime.strptime(chamado["fechamento"], "%d/%m/%Y")

    tempo_atendimento = (data_fechamento - data_abertura).days

    if tempo_atendimento > maior_tempo:
        maior_tempo = tempo_atendimento
        codigo_maior_tempo = chamado["codigo"]

    if tempo_atendimento <= 3:
        situacao = "Dentro do prazo"
        dentro_prazo.append(chamado)
    else:
        situacao = "Fora do prazo"
        fora_prazo.append(chamado)

    if chamado["prioridade"] == "Alta":
        alta.append(chamado)
    elif chamado["prioridade"] == "Média":
        media.append(chamado)
    elif chamado["prioridade"] == "Baixa":
        baixa.append(chamado)


print(f"1. Quantidade total de chamados: {len(chamados)}\n"
      f"2. Quantidade de chamados dentro do prazo: {len(dentro_prazo)}\n"
      f"3. Quantidade de chamados fora do prazo: {len(fora_prazo)}\n"
      f"4. Código do chamado com maior tempo de atendimento: {codigo_maior_tempo}\n"
      f"5. Maior tempo de atendimento encontrado: {maior_tempo} dias\n"
      f"6. Quantidade de chamados por prioridade:\n"
      f"   • Alta: {len(alta)}\n"
      f"   • Média: {len(media)}\n"
      f"   • Baixa: {len(baixa)}")
```

# Problema 1; analise de vendas por vendedor

```python
def analisar_vendas(nome, *vendas):
    print(f"Nome do vendedor: {nome}")
    print(f"Quantidade de vendas: {len(vendas)}")

    soma = sum(vendas)
    media = sum(vendas)/len(vendas)
    maior_vendas =  max(vendas)

    print(f"A soma foi: {soma} \n a media foi: {media} \n a maior venda foi: {maior_vendas}")

    if soma >= 5000:
        print(f"Parabéns {nome}, você atingiu a meta de vendas com um total de R$ {soma}")
    else:
        print(f"Infelizmente {nome}, você não atingiu a meta de vendas.")
    
```

```python
analisar_vendas("Claudio", 12000, 15000, 8000, 10000, 9000)
```
